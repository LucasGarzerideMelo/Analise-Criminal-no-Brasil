## Distribuição dos dados

### Histograma normal

```python
#@title Função
def histograma(df, coluna, bins=40):
    fig = go.Figure()
    fig.add_trace(go.Histogram(x=df[coluna], nbinsx=bins, name=coluna))

    fig.update_layout(
        title=f"Histograma da coluna {coluna}",
        xaxis_title=coluna,
        yaxis_title="Frequência",
        bargap=0.2
    )

    fig.show()
```

```python
histograma(df_2014, 'FO2014')
```

### Histograma por faixa populacionais

```python
#@title Função
def histograma_faixas_pop(df, coluna, populacao_coluna):

    bins = [0, 5000, 50000, 500000, float('inf')]
    #bins = [0, 50000, 200000, 1000000, float('inf')]  menos informativa da distribuição real
    labels = ['Pequeno', 'Médio', 'Grande', 'Muito Grande']
    df['faixa_populacional'] = pd.cut(df[populacao_coluna], bins=bins, labels=labels)


    fig = go.Figure()
    for faixa in labels:
        faixa_df = df[df['faixa_populacional'] == faixa]
        fig.add_trace(go.Histogram(
            x=faixa_df[coluna],
            name=f'Faixa: {faixa}',
            nbinsx=20
        ))

    fig.update_layout(
        title=f'Histograma de {coluna} (Agrupado por Faixas Populacionais)',
        xaxis_title=coluna,
        yaxis_title='Frequência',
        bargap=0.2,
        #xaxis_type="log",
        barmode='overlay'
    )

    fig.show()
```

### Histograma com dados normalizados por z-score

```python
#@title Função
def histograma_zscore(df, coluna):
    df['zscore'] = zscore(df[coluna])

    fig = go.Figure()
    fig.add_trace(go.Histogram(
        x=df['zscore'],
        name=f'Z-score de {coluna}',
        nbinsx=20
    ))

    fig.update_layout(
        title=f'Histograma de {coluna} (Normalizado por Z-Score)',
        xaxis_title=f'{coluna} normalizada',
        yaxis_title='Frequência',
        bargap=0.2
    )

    fig.show()
```

### Histograma com dados normalizados por min-max

```python
#@title Função
def histograma_minmax(df, coluna):
    scaler = MinMaxScaler()
    df['minmax'] = scaler.fit_transform(df[[coluna]])

    fig = go.Figure()
    fig.add_trace(go.Histogram(
        x=df['minmax'],
        name=f'Min-Max de {coluna}',
        nbinsx=20
    ))

    fig.update_layout(
        title=f'Histograma de {coluna} (Normalizado por Min-Max)',
        xaxis_title=f'{coluna} normalizada',
        yaxis_title='Frequência',
        bargap=0.2
    )

    fig.show()
```

### Histograma com transformação log

```python
#@title Função
def histograma_log_transform(df, coluna, base_log=np.e):

    if (df[coluna] <= 0).any():
        print(f"A coluna '{coluna}' contém valores <= 0. Esses valores serão ignorados.")
    dados_transformados = np.log(df[coluna][df[coluna] > 0]) / np.log(base_log)

    fig = go.Figure()
    fig.add_trace(go.Histogram(
        x=dados_transformados,
        name=f'Log-{base_log} de {coluna}',
        nbinsx=20
    ))

    fig.update_layout(
        title=f'Histograma da Coluna {coluna} (Transformação Logarítmica - Base {base_log})',
        xaxis_title=f'Log-{base_log}({coluna})',
        yaxis_title='Frequência',
        bargap=0.2
    )

    fig.show()
```

### Teste Shapiro Wilk

```python
def shapiro_test(df, coluna, nivel_significancia=0.05):

    estatistica, p_valor = shapiro(df[coluna].dropna())

    return {
        "Variável": coluna,
        "Estatística": estatistica,
        "P-Valor": p_valor,
        "Normal": p_valor > nivel_significancia,
    }
```

```python
#@title Resultados com São Paulo
resultados = {}
for coluna in df_2014.select_dtypes(include=['number']).columns:
    resultados[coluna] = shapiro_test(df_2014, coluna)


for coluna, resultado in resultados.items():
    print(f"Coluna: {coluna}")
    print(f"  Estatística: {resultado['Estatística']:.4f}")
    print(f"  P-Valor: {resultado['P-Valor']:.4f}")
    print(f"  Normal: {'Sim' if resultado['Normal'] else 'Não'}")
    print()
```

```python
#@title Resultados sem São Paulo
df_2014 = df_2014[df_2014['Municipio'] != 'Sao Paulo']

resultados = {}
for coluna in df_2014.select_dtypes(include=['number']).columns:
    resultados[coluna] = shapiro_test(df_2014, coluna)

for coluna, resultado in resultados.items():
    print(f"Coluna: {coluna}")
    print(f"  Estatística: {resultado['Estatística']:.4f}")
    print(f"  P-Valor: {resultado['P-Valor']:.4f}")
    print(f"  Normal: {'Sim' if resultado['Normal'] else 'Não'}")
    print()
```

## Entropia de Shannon

Os resultados da Entropia de Shannon confirmaram o que havia sido visto nos histogramas: os valores baixos para a entropia indicam que os dados possuem poucavariação, ou seja, estão muito agrupados.

```python
def calcular_entropia_shannon(df, coluna, bins=None):
    if bins:
        dados = pd.cut(df[coluna], bins=bins, labels=False)
    else:
        dados = df[coluna]

    frequencias = dados.value_counts()
    probabilidades = frequencias / frequencias.sum()

    return entropy(probabilidades, base=2)
```

```python
colunas_numericas = df_2014.select_dtypes(include=['number']).columns

entropias = {}
for coluna in colunas_numericas:
    entropias[coluna] = calcular_entropia_shannon(df_2014, coluna, bins=10)

print(pd.DataFrame(entropias.items(), columns=['Variável', 'Entropia']).sort_values(by='Entropia', ascending=False))
```

## Análise de outliers

Visando entender melhor os dados, fiz uma análise utilizando o método IQR para verificar a presença de outliers para analisar as distribuições após removê-los.

```python
#@title Função
def calcular_iqr(df, coluna):
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)
    IQR = Q3 - Q1
    return Q1, Q3, IQR

def identificar_outliers(df, colunas, fator=1.5):
    outliers_por_coluna = {}

    for coluna in colunas:
        Q1, Q3, IQR = calcular_iqr(df, coluna)
        limite_inferior = Q1 - fator * IQR
        limite_superior = Q3 + fator * IQR

        outliers = df[(df[coluna] < limite_inferior) | (df[coluna] > limite_superior)]
        outliers_por_coluna[coluna] = outliers

    indices_outliers_multiplos = set()
    for coluna, outliers in outliers_por_coluna.items():
        indices_outliers_multiplos.update(outliers.index)

    indices_outliers_multiplos = list(indices_outliers_multiplos)

    linhas_com_outliers_multiplos = df.loc[indices_outliers_multiplos]

    print("Linhas com outliers em mais de uma coluna:")
    print(linhas_com_outliers_multiplos)

    df_sem_outliers = df.drop(indices_outliers_multiplos)

    return df_sem_outliers, outliers_por_coluna
```

```python
colunas = ['FO2014', 'FV2014', 'LCC2014', 'HC2014', 'HCA2014', 'LCCA2014', 'LCD2014', 'VH2014', 'TH2014', 'TE2014', 'TR2014']
df_sem_outliers, outliers_por_coluna = identificar_outliers(df_2014, colunas)
```

```python
histograma(df_sem_outliers, 'FO2014')
```

```python
#@title Resultados após remoção de outliers
resultados = {}
for coluna in df_sem_outliers.select_dtypes(include=['number']).columns:
    resultados[coluna] = shapiro_test(df_sem_outliers, coluna)


for coluna, resultado in resultados.items():
    print(f"Coluna: {coluna}")
    print(f"  Estatística: {resultado['Estatística']:.4f}")
    print(f"  P-Valor: {resultado['P-Valor']:.4f}")
    print(f"  Normal: {'Sim' if resultado['Normal'] else 'Não'}")
    print()
```
