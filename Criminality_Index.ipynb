**Criação de um índice de criminalidade para análises**



O índice é calculado por ano. Para isso, criarei uma tabela para cada ano, em cada coluna referente a um crime está todas as ocorrências de tal crime no ano, e cada linha um município da região administrativa de campinas.

Para o cálculo do índice, seguirei a fórmula desse artigo (com algumas adaptações pois o número de crime analisados pelos autores foi menor): [Definition of Strategies for Crime Prevention and Combat Using Fuzzy Clustering and Formal Concept Analysis](https://www.researchgate.net/publication/320024255_Definition_of_Strategies_for_Crime_Prevention_and_Combat_Using_Fuzzy_Clustering_and_Formal_Concept_Analysis).

A tabela orignal, os pesos dados para cada crime e a fórmula orinal estão abaixo:

| Crime                       | Factor | Weight                                                                                          |
|-----------------------------|--------|-------------------------------------------------------------------------------------------------|
| Theft (victim not present)   | 1      | Theft Weight = (theft dangerousness factor) × (number of theft occurrences)                      |
| Robbery                     | 2      | Robbery Weight = (robbery dangerousness factor) × (number of robbery occurrences)                |
| Homicide                    | 3      | Homicide Weight = (homicide dangerousness factor) × (number of homicide occurrences)             |
| Drug Traffic                | 4      | Drug Traffic Weight = (traffic dangerousness factor) × (number of traffic occurrences)           |

*Table. Calculation of the Danger Factor for each crime type.*

**Fórmula:**


$\text{Danger Level} = \frac{\text{Theft.W} + \text{Robbery.W} + \text{Homicide.W} + \text{Drug.Traffic.W}}{\text{Total.Weight}}$


Como consta no artigo, os pesos para cada crime foram atribuídos por especialistas. "The weighting
of each type of crime was defined using common sense and the opinion of experts"

Como alguns dos crimes que analisaremos não constam na lista de crimes do artigo - e vice-versa -, e alguns deles possuem algumas variações no nossos dados, como homicídio, por exemplo, que nos dados do artigo é unificado, nos nossos dados possui uma separação como visto na tabela acima. Para lidar com isso, faremos algumas modificações nos pesos dados a cada crime.

Para isso, aumentaremos a escala de 1-4 para 1-6. E as explicações para cada peso seguem abaixo de cada crime e seu respectivo crime.

**Pesos mantidos:**

* Furto de veículo (peso 1)
* Furtos outros (peso 1)
* Total de roubos (peso 2)

Para esses crimes, manteremos os pesos dados no artigo, pois acreditamos não haver muita diferença entre o tipo de crime.

**Pesos atualizados por detalhamento de crime**


* Homicídio culposo por acidente de trânsito (peso 4.5)
    * Por envolver conduta negligente, imperita ou imprudente no trânsito, que podem ser causada por outros crimes, como dirigir sob efeito de álcool, acrescentamos 0.5 para ponderar a possível ocorrência - concomitante - de outro crime. Além disso, a pena é de 2 a 4 anos, enquanto a pena de homicídio culposo, de 1 a 3.

* Homicídio culposo outros (peso 4)
    * Como não há a especificação da ocorrência de outro crime concomitantemente, deixaremos 0.5 abaixo do homicídio culposo por acidente
* Homicídio doloso (peso 5)
    * Por haver intenção de realizar o ato acrescentou-se 1 ponto em relação ao homicidio culposo

**Pesos criados para crimes novos**

* Latrocínio (peso 6)
    *
* Lesão corporal culposa outras (peso 2.5)
    * Embora não haja intenção, pode colocar outra vida em risco, porém, de acordo com a lei ainda não é um crime tão grave, então o peso não está maior
* Lesão corporal culposa por acidente de trânsito (peso 3)   
    * Por involver provável imprudência na direção - além de outros crimes que podem estar envolvidos -  acrescentou-se mais 0.5
* Lesão corporal dolosa (peso 3.5)
    * Por haver intenção, adicionou-se 0.5 em relação a LCAT
* Total Estupro (peso 5)
    * Embora não haja morte como no homicídio doloso, estupro é um dos crimes mais graves e além da violação da integridade física, tem a violação da integridade psicológica que resultam em traumas que afetam as vítimas e otras pessoas.
* Tentativa de homicídio (peso 4.5)
    * Por não haver consumação do ato, retiramos 0.5 em relação ao homicídio doloso

## Definição das fórmulas para cálculo

````python
def calcular_indice_criminalidade(row, ano, pesos_dict):
    numerador = sum((row[f"{crime} {ano}"] / row[f'Populacao {ano}']) * pesos_dict[crime]  # Access dictionary using key
                    for crime in pesos_dict if f"{crime} {ano}" in row)
    denominador = sum(pesos_dict[crime] for crime in pesos_dict if f"{crime} {ano}" in row)  # Access dictionary using key
    return numerador / denominador if denominador > 0 else 0
````

````python
def normalizar_serie(serie):
    min_value = serie.min()
    max_value = serie.max()
    return (serie - min_value) / (max_value - min_value) if (max_value - min_value) > 0 else serie
````

## Método Analítico Hierárquico (AHP)



## Descrição do Método
O método AHP é uma técnica multicritério amplamente utilizada para tomar decisões complexas envolvendo múltiplos fatores. Ele organiza o problema em uma estrutura hierárquica, começando pelo objetivo principal, passando pelos critérios e subcritérios (se aplicável), até as alternativas a serem avaliadas.

### Passos Principais do AHP

---

### 1. Estruturação Hierárquica
Definir a hierarquia do problema, que inclui:
- **Objetivo Principal**: A meta que se deseja alcançar.
- **Critérios de Decisão**: Fatores que influenciam a decisão.
- **Subcritérios**: Detalhamentos dos critérios principais.
- **Alternativas**: Opções disponíveis para a decisão.

### 2. Construção da Matriz de Comparação
A matriz de comparação $A$ é formada a partir das comparações por pares entre os elementos (critérios ou alternativas). Cada entrada $a_{ij}$ da matriz representa a importância relativa do elemento $i$ em relação ao elemento $j$, utilizando a **Escala de Saaty**:

- **Escala de Saaty**:
  - 1: Importância igual.
  - 3: Importância moderada.
  - 5: Importância forte.
  - 7: Importância muito forte.
  - 9: Importância extrema.
  - Os recíprocos (e.g., $\frac{1}{3}$, $\frac{1}{5}$) são usados para relações inversas.

Além desses valores existem os intermediários: 2, 4, 6, 8.

- **Propriedades da Matriz**:
  - $a_{ij} = \frac{1}{a_{ji}}$ (reciprocidade).
  - $a_{ii} = 1$ (elementos comparados consigo mesmos).

### 3. Normalização da Matriz
Para calcular os pesos relativos:
1. **Soma-se os valores de cada coluna**:
   $$
   \text{soma}_j = \sum_{i=1}^{n} a_{ij}
   $$
2. **Dividi cada elemento $a_{ij}$ pelo total da coluna correspondente**:
   $$
   a'_{ij} = \frac{a_{ij}}{\text{soma}_j}
   $$
   Isso transforma os valores de cada coluna em frações que somam 1.

### 4. Cálculo do Vetor de Prioridade (Pesos Relativos)
Os pesos relativos $w$ são calculados como a média das linhas normalizadas:
$$
w_i = \frac{1}{n} \sum_{j=1}^{n} a'_{ij}
$$
O vetor resultante $w$ representa os pesos relativos de cada elemento.

### 5. Verificação de Consistência
Para garantir a confiabilidade das comparações, o AHP avalia a consistência da matriz utilizando o **Índice de Consistência (CI)** e a **Razão de Consistência (CR)**:

- **Índice de Consistência (CI)**:
  $$
  CI = \frac{\lambda_{\text{max}} - n}{n - 1}
  $$
  - $\lambda_{\text{max}}$: maior valor próprio da matriz original.
  - $n$: número de elementos na matriz.

- **Razão de Consistência (CR)**:
  $$
  CR = \frac{CI}{RI}
  $$
  - $RI$: Índice Aleatório de Consistência (valor tabelado).

Se $CR \leq 0.1$, a matriz é considerada **consistente**. Caso contrário, as comparações devem ser revisadas.


#### Implementação com `ahpy`
O código utiliza a biblioteca `ahpy` para automatizar:
- **Construção da Matriz de Comparação**: Definida através de dicionários que especificam as comparações por pares.
- **Cálculo dos Pesos Relativos**: Normalização da matriz e cálculo dos vetores próprios.
- **Verificação de Consistência**: Cálculo automático do índice de consistência (CR). Se $CR > 0.1$, as comparações devem ser revisadas.
